```python
# Import libraries
import pandas as pd 
import sqlite3
import matplotlib.pyplot as plt
```

```python
conn = sqlite3.connect('Level3_final_project_library.db') 
```

```python
def run (query):
    return pd.read_sql_query(query, conn)
```

```python
run("""
SELECT name 
FROM sqlite_master 
WHERE type='table';
""")
```

```python
run("""
SELECT * 
FROM members;
""")
```

```python
run("""
SELECT * 
FROM books;
""")
```

```python
run("""
SELECT * 
FROM checkouts;
""")
```

```python
run("""
SELECT first_name , COUNT(book_id) AS Borrowing_count 
FROM members m 
JOIN checkouts ch 
ON m.member_id = ch.member_id 
GROUP BY first_name;
""")
```

```python
run("""SELECT title , author 
FROM books 
WHERE author LIKE 'A%' ;""")
```

```python
run("""
SELECT title , COUNT(b.book_id) AS Borrowing_count
FROM books b 
JOIN checkouts ch ON 
b.book_id = ch.book_id 
GROUP BY title 
ORDER BY Borrowing_count DESC 
LIMIT 5;
""")
```

```python
run("""
SELECT first_name , COUNT(book_id) AS Borrowing_count 
FROM members m 
JOIN checkouts ch 
ON m.member_id = ch.member_id 
GROUP BY first_name
ORDER BY Borrowing_count DESC 
LIMIT 10 ;
""")
```

```python
run("""
SELECT first_name , COUNT(book_id) AS Borrowing_count , neighborhood , checkout_date
FROM members m
JOIN checkouts ch 
ON m.member_id = ch.member_id 
WHERE neighborhood ='Maadi'
GROUP BY first_name
ORDER BY checkout_date DESC 
LIMIT 10 OFFSET 11
""")
```
